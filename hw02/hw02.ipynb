Мои предсказания и объяснения:

1. {'A' : 2, 'B' : 1 } / Почему: переменная c идет по циклу, и с помощью словаря-накопителя .get(c, 0) считается количество букв в списке: если буквы еще нет в словаре, она добавляется со значением 1, а если уже есть к ее значению прибавляется +1. 
2. [0, 1, 1] / Почему: функция вызывается два раза подряд и каждый раз через .append(1), добавляя по одной единице в конец одного и того же исходного списка a.
3. 2 / Почему: фигурные скобки {} c циклом внутри создают множество(set), а метод lower() приводит слова к нижнему регистру, так как множество удаляет дубликаты в итоге остается только два уникальных слова {"almaty" и "astana"}, функия print(len(s)) выводит количество слов во множестве 
4. {'ab' : 2, 'abc' : 1} / Почему: {k: len(k)} cоздает словарь, где строки это ключ, а длина этих строк значения,дубликат "ab" перезаписывает свой ключ и остается две пары.
5. Первая строчка : 5, вторая None / Почему: сначала выполняется функция и выводится 5, так как нет return, она возвращает None, которое затем печатается

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.

````python
# База
def to_num(s):
    cleaned = str(s).strip().replace(' ', '')
    return float(cleaned)
````

````python
print(to_num(4))
print(to_num(56.67))
````

````python
def clean_city(s):
    s_clean = str(s).strip().lower()
    city_map = {
        'almaty': 'Алматы',
        'astana': 'Астана',
        'shymkent': 'Шымкент',
        'karaganda': 'Караганда',
        'aktobe': 'Актобе',
        'atyrau': 'Атырау'
    }
    return city_map.get(s_clean, s_clean.title())
````

````python
print(clean_city('алматы'))
````

````python
import csv

categories = []
with open('orders.csv', mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        categories.append(row['category'])

cat_counts = {}
for cat in categories:
    cat_counts[cat] = cat_counts.get(cat, 0) + 1

sorted_cats = dict(sorted(cat_counts.items(), key=lambda x: x[1], reverse=True))
print(sorted_cats)
````

````python
import csv

amounts = []
statuses = []
order_dates = []
with open('orders.csv', mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        amounts.append(row['amount'])
        statuses.append(row['status'])
        order_dates.append(row['order_date'])
````

````python
cat_stats = {}
for cat, amt, stat in zip(categories, amounts, statuses):
    if stat == 'оплачен':
        amt_num = to_num(amt)
        if cat not in cat_stats:
            cat_stats[cat] = [amt_num, 1]
        else:
            cat_stats[cat][0] += amt_num
            cat_stats[cat][1] += 1
for cat, (revenue, count) in cat_stats.items():
    avg_check = revenue / count
    print(f"Категория: {cat:<12} | Выручка: {revenue:>15,.2f} | Средний чек: {avg_check:>12,.2f}")
    
````

````python
month_counts = {}
for d in order_dates:
    month = d[:7]
    month_counts[month] = month_counts.get(month, 0) + 1
for month, count in month_counts.items():
    print(f"Месяц: {month} | Количество заказов {count:>15,}")
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
month_revenue = {}
for date, amt, stat in zip(order_dates, amounts, statuses):
    if stat == 'оплачен':
        month = date[:7]
        amt_num = to_num(amt)
        month_revenue[month] = month_revenue.get(month, 0) + amt_num
rev_sep = month_revenue['2026-09']
rev_oct = month_revenue['2026-10']
pct_change = ((rev_oct - rev_sep) / rev_sep) * 100
print(f"Выручка за сентябрь: {rev_sep:>15,.2f}")
print(f"Выручка за октябрь:  {rev_oct:>15,.2f}")
print(f"Изменение октября к сентябрю: {pct_change:+.2f}%")
````

````python
import csv

cities = []
with open('orders.csv', mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        cities.append(row['city'])
        
city_stats = {}
for c_raw, stat in zip(cities, statuses):
    c = clean_city(c_raw)
    if c not in city_stats:
        city_stats[c] = [0, 0]
    city_stats[c][1] += 1
    if stat == 'отменён':
        city_stats[c][0] += 1
for city, (canceled, total) in city_stats.items():
    cancel_rate = (canceled / total) * 100
    print(f"Город: {city:<12} | Отменено: {cancel_rate:>5.1f}%")
````

````python
import csv

id_cust = []
with open('orders.csv', mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        id_cust.append(row['customer_id'])
````

````python
customer_revenues = {}
for id, amt, stat in zip(id_cust,amounts, statuses):
    if stat == 'оплачен':
        amt_num = to_num(amt)
        customer_revenues[id] = customer_revenues.get(id, 0) + amt_num
sorted_customers = sorted(customer_revenues.items(), key=lambda x: x[1], reverse=True)
top_5 = sorted_customers[:5]
for c_id, rev in top_5:
    print(f"Customer ID: {c_id:<10} | Выручка: {rev:>15,.2f}")
    
````

ТОП 1 это тестовый чек, который возможно забыли удалить разработчики или аналитики.
У реальных клиентов выручка от 191-258 тыс. Этот чек искажает абсолютно все бизнес метрики

````python
city_customers = {}
for c_raw, c_id in zip(cities, id_cust):
    c = clean_city(c_raw)
    if c not in city_customers:
        city_customers[c] = set()
    city_customers[c].add(c_id)
for city, customers in city_customers.items():
    unique_count = len(customers)
    print(f"Город: {city:<12} | Уникальных клиентов: {unique_count:>4}")    
````
